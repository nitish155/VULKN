```python
import pandas as pd

df = pd.read_csv("datasets/Titanic-Dataset.csv")

print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())

df.head()
```

```python
print("Null count per column:")
print(df.isnull().sum())

print("\nTotal nulls in entire dataset:", df.isnull().sum().sum())
```

```python
survivors = df[df['Survived'] == 1]
non_survivors = df[df['Survived'] == 0]

print("Survivors:", len(survivors))
print("Did not survive:", len(non_survivors))

print("\nSurvivors over age 50:")
print(df[(df['Survived'] == 1) & (df['Age'] > 50)])
```

```python
print("Survival rate by passenger class:")
print(df.groupby('Pclass')['Survived'].mean())

print("\nCount by gender:")
print(df.groupby('Sex')['Survived'].count())

print("\nAverage age by class:")
print(df.groupby('Pclass')['Age'].mean())
```

```python
print("Age nulls BEFORE:", df['Age'].isnull().sum())

df['Age'] = df['Age'].fillna(df['Age'].median())

print("Age nulls AFTER:", df['Age'].isnull().sum())

df = df.drop(columns=['Cabin'])

print("\nShape after dropping Cabin column:", df.shape)
```

```python
class_info = pd.DataFrame({
    'Pclass': [1, 2, 3],
    'class_name': ['First Class', 'Second Class', 'Third Class'],
    'ticket_price_era': ['Expensive', 'Moderate', 'Cheap']
})

merged = pd.merge(df, class_info, on='Pclass')

print("Merged shape:", merged.shape)
print(merged[['Name', 'Pclass', 'class_name', 'Survived']].head(10))
```

```python
print("=== DATASET SUMMARY ===")
print(df.describe())

print("\n=== DATA TYPES ===")
print(df.dtypes)

print("\n=== VALUE COUNTS for Survived ===")
print(df['Survived'].value_counts())
```
